## 6.3.1 First Login & Initial Indexing

Once deployed, the Immich Web UI becomes accessible at `http://<PI_LOCAL_IP>:2283` or `http://homelab-pi.local:2283`. The first user to register automatically receives root administrative privileges.

### Storage Templates
By default, Immich preserves original filenames and organizes assets under internal UUID paths. In **Administration -> Settings -> Storage Template**, you can enable structured folder organization on your external SSD using patterns such as:
```text
{{y}}/{{MM}}/{{filename}}
```
This ensures your photos on disk remain neatly categorized by year and month even if accessed directly outside Immich.
