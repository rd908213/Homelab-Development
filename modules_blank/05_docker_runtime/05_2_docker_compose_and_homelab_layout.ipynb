## 5.2.1 Directory Structure & Volume Best Practices

A clean directory structure separates **code/configurations** (YAML files, env files) from **persistent application state** (database tables, uploaded photos).

```
/mnt/homelab_storage/           <-- Mounted External SSD (WD Blue SN5000)
├── docker_volumes/             <-- Heavy container database/app data
└── immich_library/             <-- Bulk photo/video storage

~/homelab/                      <-- Stored on Host OS / Git repo
├── stacks/
│   ├── immich/
│   │   ├── docker-compose.yml
│   │   └── .env
│   └── monitoring/
│       └── docker-compose.yml
```

### Bind Mounts vs. Named Volumes
- **Named Volumes (`volume_name:/path`)**: Managed inside Docker's internal directory (`/var/lib/docker/volumes/`). Less visible, harder to take direct file-level backups.
- **Host Bind Mounts (`/mnt/homelab_storage/...:/path`)**: Directly maps an exact directory path on your external SSD into the container. This makes inspecting files, migrating disks, and setting up file backups straightforward.

### Restart Policy Rationale: `unless-stopped`
- `restart: unless-stopped`: The container restarts automatically if the server reboots or if the process crashes, but remains stopped if you manually issue `docker compose stop`. (Preferred for all homelab services).

## 5.2.2 Step-by-Step Laboratory Guide

### Task 5.2.1: Verify Docker Compose v2
Ensure Compose v2 is installed as a Docker plugin:
```bash
docker compose version
```

### Task 5.2.2: Create Homelab Directory Structure
```bash
mkdir -p ~/homelab/stacks/test-stack
mkdir -p /mnt/homelab_storage/docker_volumes/test-stack
```

### Task 5.2.3: Deploy a Test Stack
Create a test compose file:
```bash
cat << 'EOF' > ~/homelab/stacks/test-stack/docker-compose.yml
services:
  web:
    image: nginx:alpine
    container_name: test_nginx
    restart: unless-stopped
    ports:
      - "8080:80"
    volumes:
      - /mnt/homelab_storage/docker_volumes/test-stack:/usr/share/nginx/html:ro

EOF
```
Launch and test:
```bash
cd ~/homelab/stacks/test-stack
docker compose up -d
docker compose ps
curl http://localhost:8080
docker compose down
```
