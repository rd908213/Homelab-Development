## 1.1.1 Conceptual Background & Operating System Analysis

Building a reliable Home Network-Attached Storage (NAS) and personal cloud requires choosing the right software and hardware foundation. When repurposing a single-board computer (SBC) like the Raspberry Pi 4 Model B (4GB RAM), the Operating System (OS) dictates stability, container compatibility, memory overhead, and driver support.

### Operating System Candidate Comparison

| Operating System | Base | Architecture | Recommended Use Case | Pros | Cons |
| :--- | :--- | :--- | :--- | :--- | :--- |
| **Raspberry Pi OS Lite (64-bit)** *(Recommended)* | Debian 12 (Bookworm) | `aarch64` | Dedicated Headless NAS / Docker host | Lowest memory footprint (~120MB idle), official firmware/kernel patches, optimal hardware video acceleration drivers. Leaves ~3.8GB RAM free for Immich! | Requires terminal-first workflow (no desktop GUI by default). |
| **Ubuntu Server 24.04 LTS (64-bit)** | Debian / Ubuntu | `aarch64` | General-purpose Homelab / Cloud-init | Broad industry familiarity, standardized Canonical packages, native cloud-init provisioning. | Higher background memory footprint (~350MB-400MB idle), less headroom on a 4GB board. |
| **DietPi** | Debian Minimal | `aarch64` | Ultra-lean lightweight setups | Highly tuned scripts, RAM-logging to reduce drive wear, interactive software installer. | Custom scripts can conflict with standard Docker / systemd homelab templates. |

### Technical Necessity: 64-Bit Architecture (`aarch64`)
Immich relies on PostgreSQL with the `pgvector` vector extension, as well as modern Machine Learning runtimes (PyTorch, ONNX runtime) for facial recognition and semantic photo search. These dependencies **do not support 32-bit (armhf)** architectures. You must select a 64-bit OS image.

### Storage Roles: MicroSD vs. External NVMe SSD
- **MicroSD Card (OS Boot Disk)**: MicroSD flash uses basic wear-leveling algorithms. Running high-frequency write operations (such as PostgreSQL database transactions, Redis caches, and Docker logs) directly on an SD card will quickly exhaust write endurance and corrupt the file system.
- **External NVMe SSD (Data & Docker Storage)**: The Western Digital WD Blue SN5000 NVMe SSD utilizes advanced wear leveling, TRIM support, and a high Terabytes Written (TBW) endurance rating. In our homelab design, the MicroSD card hosts only the static operating system, while **all Docker containers, database files, and photo media reside on the external SSD**.

## 1.1.2 Step-by-Step Laboratory Guide

### Task 1.1.1: Download and Configure Raspberry Pi Imager
1. Download the official **Raspberry Pi Imager** for your workstation from [raspberrypi.com/software](https://www.raspberrypi.com/software/).
2. Insert your target MicroSD card (Class 10 / A2 rated recommended) into your workstation card reader.
3. In Raspberry Pi Imager:
   - **Device**: Select *Raspberry Pi 4*.
   - **Operating System**: Choose `Raspberry Pi OS (other)` -> `Raspberry Pi OS Lite (64-bit)`.
   - **Storage**: Select your insertion card.
4. Click **Next**, and when prompted to edit OS customization settings, choose **Edit Settings**.
   - Set **Hostname**: e.g., `homelab-pi`
   - Set **Username & Password**: e.g., username `pi` and a secure password.
   - Configure wireless LAN (if Wi-Fi is mandatory, though wired Ethernet is strongly recommended for a NAS).
   - Under the **Services** tab: Enable **SSH** -> Select *Use password authentication* (or *Allow public-key authentication only*).
