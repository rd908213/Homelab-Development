## 1.1.1 Introduction & Conceptual Background

Building a reliable Home Network-Attached Storage (NAS) and personal cloud requires choosing the right software and hardware foundation. When repurposing a single-board computer (SBC) like the Raspberry Pi (Model 4B or 5), the Operating System (OS) dictates stability, container compatibility, memory overhead, and driver support.

### Operating System Candidate Comparison

| Operating System | Base | Architecture | Recommended Use Case | Pros | Cons |
| :--- | :--- | :--- | :--- | :--- | :--- |
| **Raspberry Pi OS Lite (64-bit)** *(Recommended)* | Debian 12 (Bookworm) | `aarch64` | Dedicated Headless NAS / Docker host | Lowest memory footprint (~120MB idle), official firmware/kernel patches, optimal hardware video acceleration drivers. | Requires terminal-first workflow (no desktop GUI by default). |
| **Ubuntu Server 24.04 LTS (64-bit)** | Debian / Ubuntu | `aarch64` | General-purpose Homelab / Cloud-init | Broad industry familiarity, standardized Canonical packages, native cloud-init provisioning. | Slightly higher background memory footprint (~350MB idle), slower kernel updates for Pi-specific peripherals. |
| **DietPi** | Debian Minimal | `aarch64` | Ultra-lean lightweight setups | Highly tuned scripts, RAM-logging to reduce drive wear, interactive software installer. | Custom scripts can conflict with standard Docker / systemd homelab templates. |

> ⚠️ **CRITICAL REQUIREMENT: 64-Bit Architecture (`aarch64`)**
> Immich relies on PostgreSQL with the `pgvector` vector extension, as well as modern Machine Learning runtimes (PyTorch, ONNX runtime) for facial recognition and semantic photo search. These dependencies **do not support 32-bit (armhf)** architectures. You must select a 64-bit OS image.

## 1.1.2 Step-by-Step Laboratory Guide

### Task 1.1.1: Download and Configure Raspberry Pi Imager
1. Download the official **Raspberry Pi Imager** for your workstation (Windows/macOS/Linux) from [raspberrypi.com/software](https://www.raspberrypi.com/software/).
2. Insert your target MicroSD card (Class 10 / A2 rated recommended) or target USB drive into your workstation.
3. In Raspberry Pi Imager:
   - **Device**: Select your Raspberry Pi model (e.g., *Raspberry Pi 4* or *Raspberry Pi 5*).
   - **Operating System**: Choose `Raspberry Pi OS (other)` -> `Raspberry Pi OS Lite (64-bit)`.
   - **Storage**: Select your insertion drive.
4. Click **Next**, and when prompted to edit OS customization settings, choose **Edit Settings**.
   - Set hostname: e.g. `homelab-pi`
   - Set username & secure password: e.g. `piadmin`
   - Configure wireless LAN (if Wi-Fi is mandatory, though wired Ethernet is strongly recommended for a NAS).
   - Enable SSH -> Select *Allow public-key authentication only* or *Use password authentication* for initial bootstrap.
