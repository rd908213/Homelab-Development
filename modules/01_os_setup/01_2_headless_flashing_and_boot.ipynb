## 1.2.1 Conceptual Background: Headless Infrastructure

A "headless" server operates without a dedicated monitor, keyboard, or mouse attached. In enterprise server management and homelab practice, headless deployments are the standard because servers live in network closets, racks, or utility shelves.

### Power Supply Delivery Considerations
When attaching external SSDs to a Raspberry Pi:
- **Raspberry Pi 4B**: Requires a stable 5V / 3A (15W) USB-C power supply. Total power delivery across all 4 USB ports is limited to 1.2A. Heavy SSD write activity can cause under-voltage brownouts unless a powered USB hub or powered enclosure is utilized.
- Always use the official 5.1V / 3.0A power supply or equivalent high-grade adapter with 18AWG wiring to prevent voltage drop.

## 1.2.2 Step-by-Step Laboratory Guide

### Task 1.2.1: Flash the Image and Verify Write Integrity
1. In Raspberry Pi Imager, apply your customizations and click **Write**.
2. Allow the write and verification stages to finish. Do not unplug the card during verification.
3. Once finished, safely eject the card and insert it into your Raspberry Pi.

### Task 1.2.2: First Boot Sequence
1. Connect an Ethernet cable from your home router/switch to the Raspberry Pi RJ45 port.
2. Connect the official USB-C power supply.
3. Observe the LED behavior:
   - Red LED: Stable power indicator (must remain solid; flickering indicates voltage drops).
   - Green LED: Activity indicator (flashes rapidly during Linux kernel boot and first-time filesystem expansion).
4. Wait approximately 90 to 120 seconds for the automated first-boot partition resizing and SSH key generation to complete.
