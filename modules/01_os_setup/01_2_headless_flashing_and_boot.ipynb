## 1.2.1 Conceptual Background: Headless Infrastructure & Power Management

A "headless" server operates without a dedicated monitor, keyboard, or mouse attached. In enterprise server management and homelab practice, headless deployments are the standard because servers live in network closets, racks, or utility shelves.

### Power Delivery & Voltage Stability
- **Raspberry Pi 4B**: Requires a stable 5V / 3A (15W) USB-C power supply. Total power delivery across all 4 USB ports is limited to 1.2A (~6W shared).
- When an external NVMe SSD is attached via USB 3.0, heavy sequential write activity causes power spikes. If voltage drops below 4.63V, the Raspberry Pi will trigger under-voltage throttling, causing CPU down-clocking or USB device resets.
- Always use the official **5.1V / 3.0A USB-C power supply** with thick 18AWG power wiring.

### Under-Voltage Diagnostics: `vcgencmd get_throttled`
You can inspect the hardware throttling state at any time via SSH using:
```bash
vcgencmd get_throttled
```
- A result of `throttled=0x0` indicates perfect, stable voltage and temperatures.
- A result of `throttled=0x50000` or `0x50005` indicates an under-voltage event occurred, signaling that a stronger power supply or powered USB hub is needed.

### Automatic Filesystem Expansion
On first boot, Raspberry Pi OS automatically detects the physical size of the MicroSD card and runs an internal expansion script (`raspi-config --expand-rootfs`). This enlarges the root partition (`/`) to fill the entire card, followed by generating unique SSH host keys before opening port 22.

## 1.2.2 Step-by-Step Laboratory Guide

### Task 1.2.1: Flash the Image and Verify Write Integrity
1. In Raspberry Pi Imager, apply your customizations and click **Write**.
2. Allow the write and verification stages to finish completely. Do not remove the card during verification.
3. Once completed, safely eject the card from your workstation and insert it into your Raspberry Pi 4 MicroSD slot.

### Task 1.2.2: First Boot Sequence
1. Connect a Cat 5e / Cat 6 Ethernet cable from your home router/switch to the Raspberry Pi RJ45 port.
2. Connect the official 5.1V / 3.0A USB-C power adapter.
3. Observe the LED behavior:
   - **Red LED (PWR)**: Solid red indicates stable voltage (flickering signals voltage drop).
   - **Green LED (ACT)**: Flashes rapidly during Linux kernel boot and first-time filesystem expansion.
4. Wait approximately 90 to 120 seconds for the initial partition resizing and SSH key generation to finish.
