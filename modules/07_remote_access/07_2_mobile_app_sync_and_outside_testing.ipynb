## 7.2.1 The Mobile Backup Experience & Backup Strategy

The ultimate objective of building a personal NAS with Immich is replicating seamless, automatic photo and video backup from your mobile device without relying on commercial cloud subscriptions.

```
[ Smartphone Camera Roll ]
            | (New Photos Taken)
            v
[ Immich Mobile App ]
            | (Encrypted WireGuard Mesh via Tailscale)
            v
[ Raspberry Pi NAS ]
            |
    +-------+-------+
    |               |
[ PostgreSQL ]  [ SSD Storage ]
 (Metadata)     (Original Photos)
```

### Mobile OS Background Sync Optimization
Both iOS and Android enforce strict power-saving limits on background network activity. To ensure Immich backs up photos in the background without needing the app open:
- **Android**: Disable *Battery Optimization* for the Immich app (set to *Unrestricted*), and allow *Background Data Usage*.
- **iOS**: Enable *Background App Refresh* in iOS Settings -> Immich.

### The 3-2-1 Backup Strategy for Homelabs
A single external SSD on a Raspberry Pi is not an invincible backup. True disaster recovery requires the **3-2-1 Rule**:
- **3 copies** of your data (Primary phone photo, Raspberry Pi SSD, Offsite backup).
- **2 different media types** (NVMe SSD storage, Cloud bucket / secondary disk).
- **1 copy offsite** (e.g., using Rclone or Restic to push encrypted snapshots of `/mnt/homelab_storage` to Backblaze B2, AWS S3 Glacier, or a family member's NAS).

## 7.2.2 Step-by-Step Laboratory Guide

### Task 7.2.1: Mobile Application Setup
1. Download **Immich** from the Apple App Store or Google Play Store on your smartphone.
2. Ensure the **Tailscale app** is installed and active (connected) on your smartphone.
3. Open Immich Mobile and configure the Server Endpoint:
   ```text
   http://homelab-nas:2283
   ```
   (or `http://100.x.y.z:2283`).
4. Log in using your Immich account credentials.

### Task 7.2.2: Configure Backup Albums
1. Grant photo library access permissions (select *Full Access* for automatic backups).
2. Under **Backup Settings**:
   - Enable backup for your main Camera folder.
   - Choose whether to back up over cellular data or require Wi-Fi.

### Task 7.2.3: Verification Test: Real-World Outside Sync
1. Turn off Wi-Fi on your mobile phone to simulate being away from home.
2. Confirm Tailscale VPN shows **Connected** in iOS/Android settings.
3. Take a new test photo on your phone.
4. Open Immich Mobile and verify the photo uploads immediately to your Raspberry Pi NAS.
5. In your workstation browser, refresh Immich to view the newly ingested photo.
