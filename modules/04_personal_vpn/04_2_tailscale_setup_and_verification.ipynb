## 4.2.1 Conceptual Background: Mesh Node Authentication

When a node joins your Tailnet, it generates an asymmetric cryptographic public/private keypair. The public key is registered with the Tailscale coordination server, which distributes routing tables to your authorized nodes. Direct data traffic travels exclusively between peers.

## 4.2.2 Step-by-Step Laboratory Guide

### Task 4.2.1: Install Tailscale on the Raspberry Pi
Connect via SSH to your Raspberry Pi and execute the official installation:
```bash
curl -fsSL https://tailscale.com/install.sh | sh
```

### Task 4.2.2: Authenticate and Launch Tailscale
Launch Tailscale and request an authentication URL:
```bash
sudo tailscale up --ssh --hostname=homelab-nas
```
*Note: The `--ssh` flag optionally enables Tailscale SSH, allowing seamless browser-based or cryptographic SSH access without managing local keys.*

Copy the generated authentication link and paste it into your browser to authorize the machine in your admin console.

### Task 4.2.3: Install Client on Phone / Laptop & Test
1. Download the Tailscale app on your phone (iOS App Store / Google Play) or laptop.
2. Sign in using the same account.
3. Verify both devices show as **Connected** in the Admin Console.
4. From your laptop or phone, test connectivity:
```bash
tailscale ping homelab-nas
ssh piadmin@homelab-nas
```
