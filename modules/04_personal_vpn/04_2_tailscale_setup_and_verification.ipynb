## 4.2.1 Conceptual Background: Mesh Node Authentication & MagicDNS

When a node joins your Tailnet, it generates an asymmetric cryptographic public/private keypair. The public key is registered with the Tailscale coordination server, which distributes routing tables to your authorized nodes. Direct data traffic travels exclusively between peers over WireGuard.

### MagicDNS Architecture
Tailscale assigns every machine an internal Carrier-Grade IPv4 address (`100.x.y.z`) and an IPv6 address. With **MagicDNS enabled**, Tailscale runs an internal resolver so you never have to remember raw IP addresses:
- Node Name: `homelab-nas`
- Full FQDN: `homelab-nas.your-tailnet.ts.net`
- Any authorized device on your account can reach the Pi simply by browsing to `http://homelab-nas:2283`!

### Tailscale SSH
Passing the `--ssh` flag configures Tailscale to authenticate SSH sessions based on your Tailscale identity. You can connect securely from any peer without needing to manage local authorized_keys files.

## 4.2.2 Step-by-Step Laboratory Guide

### Task 4.2.1: Install Tailscale on the Raspberry Pi
Connect via SSH to your Raspberry Pi and execute the official installation:
```bash
curl -fsSL https://tailscale.com/install.sh | sh
```

### Task 4.2.2: Authenticate and Launch Tailscale
Launch Tailscale and request an authentication URL:
```bash
sudo tailscale up --ssh --hostname=homelab-nas
```
Copy the generated authentication link and paste it into your workstation browser to authorize the machine in your admin console.

### Task 4.2.3: Install Client on Phone / Laptop & Test Connectivity
1. Download the **Tailscale** client on your mobile phone (iOS App Store / Google Play) and/or on your workstation laptop/desktop.
2. Sign in using the same identity provider account and toggle the Tailscale connection **On**.
3. Verify both `homelab-nas` and your secondary device show as **Connected** in the Admin Console.
4. Test peer-to-peer connectivity using any of the methods below:

#### Option A: Pinging from a Mobile Phone (iOS / Android Tailscale App)
The Tailscale mobile app includes a built-in WireGuard ping tool:
1. Open the **Tailscale** app on your phone (for a true remote test, turn **off Wi-Fi** so your phone uses cellular data).
2. Tap on **`homelab-nas`** in your device list to open its machine details drawer.
3. Tap **Ping `homelab-nas`** (or **Ping**). The app will send up to 10 WireGuard discovery pings and display the round-trip latency (ms) and whether it established a **Direct** peer-to-peer connection or routed via a **DERP** relay.

#### Option B: Pinging Your Phone (or Laptop) from the Raspberry Pi Terminal
With Tailscale active on your phone or laptop, run `tailscale status` on your Raspberry Pi to list all peers on your Tailnet, then run `tailscale ping` targeting your phone's hostname or `100.x.y.z` Tailscale IP:
```bash
tailscale status
tailscale ping <your-phone-or-laptop-hostname>
```

#### Option C: Pinging `homelab-nas` from a Laptop / Desktop Terminal
If you installed the Tailscale client on your laptop or desktop workstation, open PowerShell or Terminal and run:
```bash
tailscale status
tailscale ping homelab-nas
ssh pi@homelab-nas
```
