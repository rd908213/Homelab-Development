## 4.1.1 The Homelab Remote Access Challenge

Historically, accessing a home server from outside required:
1. Setting up Dynamic DNS (DDNS).
2. Opening ports (port forwarding) on the home router.
3. Exposing your server's ports to malicious automated internet crawlers and botnets.

Furthermore, many modern ISPs utilize **CGNAT (Carrier-Grade NAT)**, meaning your router shares a public IP with hundreds of subscribers. In a CGNAT environment, port forwarding on your router is mathematically impossible.

```
Traditional (High Risk & Fails under CGNAT):
[Internet / Botnets] ---> [Router Port Forward: 2283] ---> [Immich Server]

Mesh VPN / Tailscale (Zero Open Ports, Secure WireGuard Mesh):
[Mobile Phone Client] <===============================> [Raspberry Pi NAS]
                      (Direct Encrypted P2P WireGuard Tunnel)
```

### NAT Traversal Mechanics (STUN & DERP)
- **STUN (Session Traversal Utilities for NAT)**: Devices discover their public IP and port mappings. When two peers want to communicate, they use UDP hole punching to establish a direct peer-to-peer tunnel through both firewalls.
- **DERP (Designated Encrypted Relay for Packets)**: If symmetric firewalls prevent direct peer-to-peer hole punching, Tailscale routes encrypted traffic through the nearest geographically distributed DERP relay server. Even over DERP relays, traffic remains end-to-end encrypted; the relay cannot inspect payloads.

### Why WireGuard Beats Legacy OpenVPN on ARM
WireGuard runs directly in the Linux kernel space and utilizes modern, efficient cryptographic primitives: **ChaCha20-Poly1305** for symmetric encryption and **Curve25519** for key exchange. This architecture drastically reduces CPU utilization on the Raspberry Pi 4 compared to legacy OpenVPN, yielding near-line-rate network throughput.

### Task 4.1.2: Check CGNAT Status on Your Home Network
Determine whether your home connection is behind CGNAT:
1. Check the WAN IP displayed in your router's status page.
2. If your WAN IP falls within `100.64.0.0/10` (RFC 6598: `100.64.0.0` to `100.127.255.255`), your ISP is using CGNAT.
3. Compare that IP to what `curl ifconfig.me` returns.
