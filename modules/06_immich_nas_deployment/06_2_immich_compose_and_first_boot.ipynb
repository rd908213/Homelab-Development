## 6.2.1 Multi-Container Orchestration & Health Checks

During first boot, Immich coordinates four simultaneous containers. The `immich-server` container uses Docker Compose health checks to wait for both `redis` and `database` to report healthy status before executing database schema migrations.

### Database Backup Strategy
All photo metadata, facial bounding boxes, user accounts, and album structures reside in PostgreSQL. Backing up the `/mnt/homelab_storage/docker_volumes/immich_postgres` directory or configuring an automated `pg_dump` backup is essential for complete disaster recovery.

## 6.2.2 Step-by-Step Laboratory Guide

### Task 6.2.1: Pull Container Images
Pull the ARM64 images to your Raspberry Pi:
```bash
cd ~/homelab/stacks/immich
docker compose pull
```

### Task 6.2.2: Launch the Stack in Detached Mode
```bash
docker compose up -d
```

### Task 6.2.3: Monitor Startup Logs & Health Status
Follow container initialization logs:
```bash
docker compose logs -f immich-server
```
Check health status across all containers:
```bash
docker compose ps
```
All containers (`immich_server`, `immich_machine_learning`, `immich_redis`, `immich_postgres`) should transition to `running` / `healthy` status.
