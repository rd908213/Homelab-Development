## 6.1.1 Immich Microservices Architecture

Immich is a modern, high-performance self-hosted photo and video backup solution designed as an open-source alternative to Google Photos and Apple iCloud.

```
+-------------------------------------------------------------+
|                      Immich Client                          |
|         (Web Browser, iOS App, Android Mobile App)          |
+------------------------------+------------------------------+
                               | Port 2283
                               v
+-------------------------------------------------------------+
|                     immich-server                           |
|        (Core API, OAuth, Asset Processing Pipeline)         |
+---------------+----------------+--------------------+-------+
                |                |                    |
                v                v                    v
     +--------------------+  +-------+  +--------------------+
     |   immich-postgres  |  | Redis |  |  immich-machine-   |
     |  (Data + pgvector) |  | (Queue|  |      learning      |
     |                    |  | Cache)|  | (CLIP & Facial Rec)|
     +--------------------+  +-------+  +--------------------+
                |
                v
    [ /mnt/homelab_storage/immich_library/ ] (SSD)
```

### Raspberry Pi Resource Optimization Tips
1. **Machine Learning Concurrency**: Running facial recognition and smart search models can saturate ARM cores. Set `MACHINE_LEARNING_WORKERS=1` in your `.env`.
2. **PostgreSQL Tuning**: PostgreSQL with vector indexes requires sufficient shared memory buffer.

## 6.1.2 Step-by-Step Laboratory Guide

### Task 6.1.1: Prepare Deployment Directory
```bash
mkdir -p ~/homelab/stacks/immich
mkdir -p /mnt/homelab_storage/immich_library
mkdir -p /mnt/homelab_storage/docker_volumes/immich_postgres
cd ~/homelab/stacks/immich
```

### Task 6.1.2: Download Configuration Template
Download official `.env` and `docker-compose.yml`:
```bash
wget -O docker-compose.yml https://github.com/immich-app/immich/releases/latest/download/docker-compose.yml
wget -O .env https://github.com/immich-app/immich/releases/latest/download/example.env
```

### Task 6.1.3: Customize `.env` for External SSD
Edit `.env`:
```bash
nano .env
```
Ensure key variables are set:
```text
UPLOAD_LOCATION=/mnt/homelab_storage/immich_library
DB_DATA_LOCATION=/mnt/homelab_storage/docker_volumes/immich_postgres
IMMICH_VERSION=release
DB_PASSWORD=SetAStrongRandomPasswordHere!
MACHINE_LEARNING_WORKERS=1
```
