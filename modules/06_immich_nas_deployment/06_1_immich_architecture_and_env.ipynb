## 6.1.1 Immich Microservices Architecture

Immich is a modern, high-performance self-hosted photo and video backup solution designed as an open-source alternative to Google Photos and Apple iCloud.

```
+-------------------------------------------------------------+
|                      Immich Client                          |
|         (Web Browser, iOS App, Android Mobile App)          |
+------------------------------+------------------------------+
                               | Port 2283
                               v
+-------------------------------------------------------------+
|                     immich-server                           |
|        (Core API, OAuth, Asset Processing Pipeline)         |
+---------------+----------------+--------------------+-------+
                |                |                    |
                v                v                    v
     +--------------------+  +-------+  +--------------------+
     |   immich-postgres  |  | Redis |  |  immich-machine-   |
     |  (Data + pgvector) |  | (Queue|  |      learning      |
     |                    |  | Cache)|  | (CLIP & Facial Rec)|
     +--------------------+  +-------+  +--------------------+
                |
                v
    [ /mnt/homelab_storage/immich_library/ ] (WD Blue SN5000 SSD)
```

### Microservices Breakdown
1. **`immich-server`**: Core REST API, handles authentication, routes uploads, and controls business logic.
2. **`immich-postgres`**: Stores asset metadata, albums, and uses `pgvector` for semantic image embeddings.
3. **`immich-redis`**: In-memory job queue coordinator (thumbnail generation, facial detection jobs).
4. **`immich-machine-learning`**: Runs facial recognition models and CLIP neural networks for smart search.

### Vector Search with `pgvector`
The `pgvector` PostgreSQL extension stores multi-dimensional vectors generated by the CLIP machine learning model. When you search for "sunset at the beach", Immich converts your query into a vector and finds photos with mathematically similar embeddings, without needing any manual tagging.

### Specific Memory Tuning for Raspberry Pi 4 (4GB RAM)
- Set `MACHINE_LEARNING_WORKERS=1` in your `.env`.
- Stick to lightweight models (`ViT-B-32__openai` or `tiny-vit`). This ensures face recognition and CLIP indexing complete smoothly without exhausting memory.

## 6.1.2 Step-by-Step Laboratory Guide

### Task 6.1.1: Prepare Deployment Directory on External SSD
```bash
mkdir -p ~/homelab/stacks/immich
mkdir -p /mnt/homelab_storage/immich_library
mkdir -p /mnt/homelab_storage/docker_volumes/immich_postgres
cd ~/homelab/stacks/immich
```

### Task 6.1.2: Download Configuration Template
Download official `.env` and `docker-compose.yml`:
```bash
wget -O docker-compose.yml https://github.com/immich-app/immich/releases/latest/download/docker-compose.yml
wget -O .env https://github.com/immich-app/immich/releases/latest/download/example.env
```

### Task 6.1.3: Customize `.env` for External SSD and 4GB RAM Limits
Edit `.env`:
```bash
nano .env
```
Ensure key variables are set:
```text
UPLOAD_LOCATION=/mnt/homelab_storage/immich_library
DB_DATA_LOCATION=/mnt/homelab_storage/docker_volumes/immich_postgres
IMMICH_VERSION=release
DB_PASSWORD=SetAStrongRandomPasswordHere!
MACHINE_LEARNING_WORKERS=1
```
